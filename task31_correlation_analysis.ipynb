```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    "Hours_Studied": [1, 2, 3, 4, 5, 6, 7, 8],
    "Marks": [35, 40, 50, 55, 65, 70, 80, 90],
    "Sleep_Hours": [8, 7, 8, 6, 7, 6, 5, 6],
    "Attendance": [60, 65, 70, 75, 80, 85, 90, 95]
}
df = pd.DataFrame(data)
print(df)
```

```python
print(df.info())
```

```python
correlation_matrix = df.corr()
print(correlation_matrix)
```

```python
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.scatterplot(data=df, x="Hours_Studied", y="Marks")
plt.title("Study Hours vs Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.show()
```

```python
sns.scatterplot(data=df, x="Attendance", y="Marks")
plt.title("Attendance vs Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Marks")
plt.show()
```

```python
sns.scatterplot(data=df, x="Sleep_Hours", y="Marks")
plt.title("Sleep Hours vs Marks")
plt.xlabel("Sleep Hours")
plt.ylabel("Marks")
plt.show()
```

```python
study_marks_corr = df["Hours_Studied"].corr(df["Marks"])
print("Correlation between Study Hours and Marks:", round(study_marks_corr, 2))
```

```python
attendance_marks_corr = df["Attendance"].corr(df["Marks"])
print("Correlation between Attendance and Marks:", round(attendance_marks_corr, 2))
```

```python
correlation_pairs = correlation_matrix.where(
    correlation_matrix.abs() < 1
).stack().sort_values(key=abs, ascending=False)

print("Correlation pairs:")
print(correlation_pairs)
```

```python
print("Correlation Analysis Results:")
print(f"Study Hours vs Marks: {study_marks_corr:.2f}")
print(f"Attendance vs Marks: {attendance_marks_corr:.2f}")

if study_marks_corr > 0:
    print("Study Hours and Marks have a positive correlation.")
elif study_marks_corr < 0:
    print("Study Hours and Marks have a negative correlation.")
else:
    print("Study Hours and Marks have no linear correlation.")

print("Note: Correlation does not imply causation.")
```
